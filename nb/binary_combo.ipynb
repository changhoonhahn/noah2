with enough communities and zipcodes that can be used as base distributions. 

**Enough is defined as at least 400 communities and at least 25 unique zipcodes**

# read data
read data set of all CRS participants

```python
# read CRS participant data
DNoah = D.Noah2()
fema = DNoah._read_data_full()
is_participant = DNoah._participants(fema)
fema = fema[is_participant]
```

```python
metro = ((fema['RUCA1'] == 1))# | (fema['RUCA1'] == 2) | (fema['RUCA1'] == 3)) 
micro = ((fema['RUCA1'] == 4) | (fema['RUCA1'] == 5) | (fema['RUCA1'] == 6))
small = ((fema['RUCA1'] == 7) | (fema['RUCA1'] == 8) | (fema['RUCA1'] == 9))
rural = ((fema['RUCA1'] == 10))
```

```python
arch_dict =  {'rural': rural, 'small': small, 'micro': micro, 'metro': metro, 
              'all': np.ones(len(fema)).astype(bool)}
```

```python
binary_activities = ['c360', 'c520', 'c530', 'c540', 'c610', 'c620', 'c630']

binary_data = np.array([np.array(fema[col]) for col in binary_activities]).T
binary_data = (binary_data > 0).astype(int)

binary_act_codes = np.zeros(len(fema))
for i in range(binary_data.shape[1]): 
    binary_act_codes += 2**i * binary_data[:,i]
```

```python
def uncode(num): 
    act = np.zeros(len(binary_activities)).astype(int)
    while num > 0: 
        ind = int(np.floor(np.log(num)/np.log(2)))
        num = num - 2**ind
        act[ind] = 1
    return act
```

```python
def find_base_combo(archetype, ncomm_cut=400, nzip_cut=25): 
    arch = arch_dict[archetype] # archetype
    
    u_act_codes = np.unique(binary_act_codes[arch])
    
    base_codes = []
    for code in u_act_codes:
        is_code = (binary_act_codes == code) & arch
        
        ncomm = np.sum(is_code)
        nzip = len(np.unique(fema['zipcode'][is_code]))
        
        if (nzip >= nzip_cut) and (ncomm >= ncomm_cut): 
            print('%i: Ncomm = %i, Nzip = %i' % (code, ncomm, nzip))
            base_codes.append(int(code))
    return np.array(base_codes)
```

# base distribution for each archetype

```python
for arch in ['metro', 'micro', 'small', 'rural']: 
    print(arch)
    print(find_base_combo(arch, ncomm_cut=400, nzip_cut=25))
    print()
```
